```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
data = pd.read_csv("adult.csv")
data.shape
```

```python
data.head()
```

```python
column_names = ["age", "workclass", "final_weight", "education", "education_num","marital_status","occupation","relationship","race","sex","capital_gain","capital_loss","hours_per_week","native_country","income"]

data.columns = column_names
data 
```

```python
data.isnull().sum()
```

```python
missing_values = (data == ' ?').sum()
data.replace('?', np.nan, inplace=True)
data.isnull().sum()   
```

```python
sns.heatmap(data.isna(), cbar=True)
plt.show()  
```

```python
sns.countplot(x='income', data=data)
plt.title('Income Distribution')
plt.show()
```

```python
fig = plt.figure(figsize=(15,4))
sns.lineplot(data=data, x='age', y='final_weight', hue='income')
plt.show()
```

```python
sns.countplot(x='sex', data=data)
plt.title('Gender Distribution')
plt.show()
```

```python
sns.countplot(x='income', hue='sex',data=data)
plt.title('Income Distribution Based on Gender')
plt.show()
```

```python
fig = plt.figure(figsize=(15,4))
plt.title('Final Weight Distribution Based on Workclass')
sns.barplot(x = 'workclass', y = 'final_weight', hue = 'income', data = data)
plt.plot()
```

```python
fig = plt.figure(figsize=(14,4))
sns.countplot(x='workclass', data=data)
plt.title('Workclass Distribution')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.histplot(data['age'], bins=10, kde=True)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(18, 6))
sns.countplot(x='education', data=data)
plt.title('Education Distribution')
plt.show()
```

```python
plt.figure(figsize=(18, 6))
sns.scatterplot(data=data, x = 'hours_per_week', y = 'income')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(x = 'race', data=data)
plt.title('Race Distribution')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='income', hue='race',data=data)
plt.title('Income Distribution Based on Race')
plt.show()
```

```python
plt.figure(figsize=(20, 6))
sns.countplot(x = 'education',hue = 'income', data = data)
plt.title('Income based on Education')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.histplot(data['hours_per_week'], bins=10, kde=True)
plt.title('Hours Worked Per Week Distribution')
plt.xlabel('Hours per week')
plt.ylabel('Frequency')
plt.show()
```

```python
print(data['native_country'].value_counts().values)
data['native_country'].value_counts().index
```

```python
top_native_countries = data['native_country'].value_counts().head()

plt.figure(figsize=(10, 6))
sns.barplot(x=top_native_countries.values, y=top_native_countries.index)
plt.title('Top 5 Native Countries')
plt.xlabel('Count')
plt.ylabel('Native Country')
plt.show()
```
